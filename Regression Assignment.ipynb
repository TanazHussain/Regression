## **Question 1:** What is Simple Linear Regression (SLR)? Explain its purpose.  
**Answer:**  
# Simple Linear Regression (SLR)

## **Definition**
**Simple Linear Regression (SLR)** is a **statistical technique** used to examine and model the **linear relationship** between two quantitative variables:

- **Dependent variable (Y):** The outcome or variable we want to predict or explain.  
- **Independent variable (X):** The predictor or explanatory variable used to predict changes in Y.

The relationship between X and Y is represented by the **straight-line equation**:

\[
Y = \beta_0 + \beta_1X + \varepsilon
\]

Where:
- \( \beta_0 \): Intercept (value of Y when X = 0)  
- \( \beta_1 \): Slope (change in Y for a one-unit change in X)  
- \( \varepsilon \): Random error term representing other unexplained factors  

---

## **Purpose of Simple Linear Regression**

1. **Prediction:**  
   To predict the value of the dependent variable (Y) for a given value of the independent variable (X).  
   *Example:* Predicting a student’s exam score (Y) based on hours studied (X).

2. **Understanding Relationships:**  
   To determine whether and how strongly X influences Y.  
   The slope (\( \beta_1 \)) indicates the **direction** (positive or negative) and **strength** of the relationship.

3. **Quantifying Effects:**  
   It helps quantify how much change in Y can be expected for a unit change in X.

4. **Identifying Trends:**  
   Provides a simple way to understand trends and make data-driven decisions.


## **Question 2:** What are the key assumptions of Simple Linear Regression?  
**Answer:**  
The key assumptions are:  
1. **Linearity:** The relationship between X and Y is linear.  
2. **Independence:** Observations are independent of each other.  
3. **Homoscedasticity:** The variance of residuals is constant across all levels of X.  
4. **Normality of errors:** The residuals are normally distributed.  
5. **No multicollinearity:** (Applies to multiple regression) – since SLR uses only one X, this isn’t relevant here.



## **Question 3:** Write the mathematical equation for a simple linear regression model and explain each term.  
**Answer:**  
# Mathematical Equation of a Simple Linear Regression Model

The mathematical equation for a **Simple Linear Regression (SLR)** model is given by:

\[
Y_i = \beta_0 + \beta_1 X_i + \varepsilon_i
\]

---

## **Explanation of Terms**

- **\( Y_i \)** — The *dependent variable* (or response variable) for the *i-th* observation.  
  It represents the outcome or value we are trying to predict.

- **\( X_i \)** — The *independent variable* (or predictor variable) for the *i-th* observation.  
  It represents the input or factor used to predict \( Y_i \).

- **\( \beta_0 \)** — The *intercept* of the regression line.  
  It is the expected value of \( Y \) when \( X = 0 \).  
  In other words, it represents where the regression line crosses the Y-axis.

- **\( \beta_1 \)** — The *slope* of the regression line.  
  It indicates the amount by which \( Y \) changes for a one-unit increase in \( X \).  
  - If \( \beta_1 > 0 \): There is a **positive relationship** between X and Y.  
  - If \( \beta_1 < 0 \): There is a **negative relationship** between X and Y.

- **\( \varepsilon_i \)** — The *error term* (or residual) for the *i-th* observation.  
  It represents the difference between the actual observed value and the predicted value of Y.  
  Mathematically:  
  \[
  \varepsilon_i = Y_i - (\beta_0 + \beta_1 X_i)
  \]
  The error term captures all other factors that affect Y but are not included in the model.

---

## **Interpretation**
The equation represents a **straight-line relationship** between X and Y.  
- \( \beta_0 \) determines the *starting point* of the line.  
- \( \beta_1 \) determines the *tilt or steepness* of the line.  
- The error term \( \varepsilon_i \) accounts for random variation not explained by X.

# Real-World Example of Simple Linear Regression

## **Example: Predicting House Prices Based on Size**

### **Scenario**
A real estate analyst wants to understand how the **size of a house (in square feet)** affects its **price (in dollars)**.  
They collect data on several houses, recording each home’s size (X) and corresponding price (Y).

The goal is to **predict house price (Y)** based on the **house size (X)** and to understand how strongly they are related.

---

### **Model Representation**
The relationship can be expressed using the simple linear regression equation:

\[
Y = \beta_0 + \beta_1 X + \varepsilon
\]

Where:
- \( Y \): House price (dependent variable)  
- \( X \): House size in square feet (independent variable)  
- \( \beta_0 \): Intercept (expected price when size = 0)  
- \( \beta_1 \): Slope (change in price for each additional square foot)  
- \( \varepsilon \): Random error term  

---

### **Example Calculation**
After fitting the model to data, suppose the estimated regression equation is:

\[
\hat{Y} = 50,000 + 200X
\]

**Interpretation:**
- **Intercept (\( \beta_0 = 50,000 \))**:  
  Even if a house had zero square feet (theoretical), the model predicts a base price of \$50,000.

- **Slope (\( \beta_1 = 200 \))**:  
  For every additional square foot, the predicted price of the house increases by \$200.

**Example Prediction:**
If a house is **1,500 sq. ft**, then the predicted price is:

\[
\hat{Y} = 50,000 + 200(1,500) = 350,000
\]

So, a 1,500 sq. ft home is expected to cost **\$350,000** according to the model.

---

### **Interpretation and Use**
This simple linear regression model helps:
- **Home buyers** estimate property prices based on size.  
- **Sellers** set reasonable asking prices.  
- **Real estate analysts** understand the relationship between property size and market value.

---

### **Key Takeaway**
Simple Linear Regression provides a **quantitative way to model and predict** outcomes when there’s a **linear relationship** between two variables — in this case, house size and price.




## **Question 5:** What is the method of least squares in linear regression?  
**Answer:**  
# Method of Least Squares in Linear Regression

## **Definition**

The **Method of Least Squares** is a mathematical approach used in **linear regression** to find the *best-fitting line* through a set of data points.  
It works by **minimizing the sum of the squares of the residuals** (the differences between the observed and predicted values of the dependent variable).

---

## **Mathematical Explanation**

In a simple linear regression model:

\[
Y_i = \beta_0 + \beta_1 X_i + \varepsilon_i
\]

- \( Y_i \): Actual observed value of the dependent variable  
- \( X_i \): Observed value of the independent variable  
- \( \hat{Y_i} = \beta_0 + \beta_1 X_i \): Predicted value from the regression line  
- \( \varepsilon_i = Y_i - \hat{Y_i} \): Residual (error term) for the *i-th* observation  

The **method of least squares** chooses values of \( \beta_0 \) and \( \beta_1 \) that minimize the **Sum of Squared Errors (SSE)**:

\[
SSE = \sum_{i=1}^{n} (Y_i - \hat{Y_i})^2 = \sum_{i=1}^{n} (Y_i - \beta_0 - \beta_1 X_i)^2
\]

---

## **Objective**

Find \( \beta_0 \) and \( \beta_1 \) such that the SSE is **as small as possible**.  
This ensures the fitted line is the *closest possible* to all data points in terms of squared vertical distances.

---

## **Formulas for Coefficients**

By minimizing SSE (taking partial derivatives with respect to \( \beta_0 \) and \( \beta_1 \)), we get:

\[
\beta_1 = \frac{\sum (X_i - \bar{X})(Y_i - \bar{Y})}{\sum (X_i - \bar{X})^2}
\]

\[
\beta_0 = \bar{Y} - \beta_1 \bar{X}
\]

Where:
- \( \bar{X} \) = Mean of the independent variable  
- \( \bar{Y} \) = Mean of the dependent variable  

---

## **Interpretation**

- The **slope (\( \beta_1 \))** represents the average change in Y for a one-unit change in X.  
- The **intercept (\( \beta_0 \))** represents the predicted value of Y when X = 0.  
- The **least squares line** minimizes the total squared difference between actual and predicted Y values.

---

## **Example**

Suppose we have data on:
- \( X \): Number of hours studied  
- \( Y \): Exam scores  

The least squares method finds the line that best fits the relationship between study time and exam performance by minimizing the sum of squared prediction errors.

---

## **Key Takeaway**

The **Method of Least Squares** is the foundation of linear regression.  
It provides a **systematic and optimal way** to estimate the regression line by minimizing the total squared error, ensuring the best possible linear fit to the data.



## **Question 6:** What is Logistic Regression? How does it differ from Linear Regression?  
**Answer:**  
**Logistic Regression** is used to model a **binary or categorical dependent variable** (e.g., yes/no, 0/1).  
It uses the **sigmoid (logistic) function** to output probabilities between 0 and 1.  

**Differences:**  
| Feature | Linear Regression | Logistic Regression |
|----------|-------------------|---------------------|
| Output | Continuous | Categorical (probability) |
| Function | Linear | Sigmoid (logistic) |
| Purpose | Predicts values | Classifies outcomes |



## **Question 7:** Name and briefly describe three common evaluation metrics for regression models.  
**Answer:**  
1. **Mean Absolute Error (MAE):** Average of absolute differences between actual and predicted values.  
2. **Mean Squared Error (MSE):** Average of squared differences between actual and predicted values.  
3. **R-squared (R²):** Proportion of variance in Y explained by X.



## **Question 8:** What is the purpose of the R-squared metric in regression analysis?  
**Answer:**  
# Purpose of the R-squared Metric in Regression Analysis

## **Definition**

**R-squared (R²)**, also known as the **coefficient of determination**, is a statistical measure that represents the **proportion of the variance** in the dependent variable (**Y**) that is **explained by the independent variable(s)** in a regression model.

It provides an indication of how **well the regression line fits the data**.

---

## **Mathematical Formula**

\[
R^2 = 1 - \frac{SS_{res}}{SS_{tot}}
\]

Where:

- \( SS_{res} = \sum (Y_i - \hat{Y_i})^2 \) → **Residual Sum of Squares (unexplained variance)**  
- \( SS_{tot} = \sum (Y_i - \bar{Y})^2 \) → **Total Sum of Squares (total variance in Y)**  
- \( Y_i \): Actual observed values  
- \( \hat{Y_i} \): Predicted values from the regression model  
- \( \bar{Y} \): Mean of the observed dependent variable  

---

## **Interpretation**

- \( R^2 \) measures how much of the variation in the dependent variable is **explained by the model**.  
- The value of \( R^2 \) lies between **0 and 1**:

| **R² Value** | **Interpretation** |
|---------------|--------------------|
| **0** | The model explains **none** of the variability in Y. |
| **1** | The model explains **all** the variability in Y (perfect fit). |
| **Between 0 and 1** | Indicates the **percentage of variance** in Y explained by X. For example, R² = 0.8 means 80% of the variation in Y is explained by X. |

---

## **Example**

Suppose we model the relationship between **hours studied (X)** and **exam scores (Y)** and get:

\[
R^2 = 0.85
\]

This means that **85% of the variation** in students’ exam scores can be explained by the number of hours they studied.  
The remaining **15%** is due to other factors (e.g., difficulty of the test, student’s background, etc.).

---

## **Key Points**

- A **higher R²** value generally indicates a **better fit** of the model to the data.  
- However, **a high R² doesn’t always mean the model is good** — it only measures the fit, not whether the model is appropriate or unbiased.  
- R² **does not indicate causation** between variables.  
- In multiple regression, an **Adjusted R²** is often used to account for the number of predictors in the model.

---

## **Summary**

**R-squared** helps to:
- Evaluate the **goodness of fit** of a regression model.  
- Understand how well the model explains the observed data.  
- Compare the explanatory power of different regression models.




## **Question 9:** Write Python code to fit a simple linear regression model using scikit-learn and print the slope and intercept.  
**Answer:**


```python

# Import necessary libraries
import numpy as np
from sklearn.linear_model import LinearRegression

# Sample data
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)  # Independent variable
y = np.array([2, 4, 5, 4, 5])                 # Dependent variable

# Create and fit the model
model = LinearRegression()
model.fit(X, y)

# Print slope (coefficient) and intercept
print("Slope (β₁):", model.coef_[0])
print("Intercept (β₀):", model.intercept_)
```


## **Question 10:** How do you interpret the coefficients in a simple linear regression model?  
**Answer:**  
# Interpreting the Coefficients in a Simple Linear Regression Model

## **Mathematical Model**

A **Simple Linear Regression (SLR)** model is represented as:

\[
Y = \beta_0 + \beta_1 X + \varepsilon
\]

Where:
- \( Y \): Dependent (response) variable  
- \( X \): Independent (predictor) variable  
- \( \beta_0 \): Intercept (constant term)  
- \( \beta_1 \): Slope (regression coefficient)  
- \( \varepsilon \): Error term (difference between actual and predicted Y)

---

## **1. Intercept (\( \beta_0 \))**

- The **intercept** represents the **predicted value of Y when X = 0**.  
- It is the point where the regression line crosses the **Y-axis**.  
- In some contexts, it may not have a meaningful interpretation if X = 0 is outside the range of the observed data.

**Example:**  
If the regression equation is:
\[
\hat{Y} = 40 + 5X
\]
Then \( \beta_0 = 40 \) means that when \( X = 0 \), the predicted value of Y is 40.

---

## **2. Slope (\( \beta_1 \))**

- The **slope coefficient** represents the **average change in Y** for a **one-unit increase in X**, assuming all other factors are constant.  
- It shows the **direction** and **magnitude** of the relationship:
  - If \( \beta_1 > 0 \): Positive relationship (Y increases as X increases)  
  - If \( \beta_1 < 0 \): Negative relationship (Y decreases as X increases)

**Example:**  
Using the same equation:
\[
\hat{Y} = 40 + 5X
\]
- \( \beta_1 = 5 \): For every additional unit increase in X, Y increases by 5 units on average.

---

## **3. Putting It Together**

In the equation:
\[
\hat{Y} = \beta_0 + \beta_1 X
\]
- The regression line **predicts the expected value of Y** given a specific value of X.  
- \( \beta_0 \) sets the **starting point (baseline)**.  
- \( \beta_1 \) determines the **rate of change** or **sensitivity** of Y to X.

---

## **Example Interpretation**

Suppose we model the relationship between:
- **X:** Hours studied  
- **Y:** Exam score  

and obtain the regression equation:

\[
\hat{Y} = 50 + 10X
\]

**Interpretation:**
- \( \beta_0 = 50 \): A student who studies 0 hours is expected to score 50 points.  
- \( \beta_1 = 10 \): For each additional hour studied, the student’s exam score increases by 10 points on average.

---

## **Key Takeaways**

- **Intercept (\( \beta_0 \))** → Predicted value of Y when X = 0.  
- **Slope (\( \beta_1 \))** → Change in Y for a one-unit change in X.  
- Together, they define the **line of best fit** that summarizes the linear relationship between X and Y.
